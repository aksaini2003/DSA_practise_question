```python
#code for finding the gcd of 2 numbers 

def gcd(a,b):
    hcf=0
    if a< b : 
        for i in range(a,0,-1):
            if b%i==0 and a%i==0: 
                hcf=i 
                break 
    else: 
        for i in range(b,0,-1):
            if a%i==0 and b%i==0:
                hcf=i 
                break
    return hcf

print(gcd(12,60))

def optimal_gcd(a,b):
    while b > 0 :
        a,b=b,a%b 
    return a 


print(optimal_gcd(120, 60))  # Output: 12
```

```python
#now lets see the left and right rotation in the array 
#lets see both rotations 

def array_rotation(arr,type):
    if type=='left':
        start=arr[0] 
        for i in range(1,len(arr)):
            arr[i-1]=arr[i] 
        arr[i]=start  
        
    elif type=='right': 
        end=arr[len(arr)-1] 
        for i in range(len(arr)-1,0,-1): 
            arr[i]=arr[i-1] 
        arr[0]=end 
    
    else: 
        return "please enter a valid type"
    return arr 
print(array_rotation([1,24,5,6,6,7],'right'))
```

```python
#finding the element which occurred only once 
def single_occurrence(arr):
    mi=min(arr) 
    ans=arr[0] 
    for i in arr[1:]:
        ans=ans^i 
        
    if ans==0 and mi==0:
        return 0 
    elif ans==0 and mi!=0: 
        return "no single occurrence element found"
    else: 
        return ans 
print(single_occurrence([1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,0])  )
```

```python
#finding majority element occurring > len(arr)//2 
#lets see the simplest approach for this 
def find_majority(arr):
    count_dict=dict() 
    for i in range(len(arr)):
        if count_dict.get(arr[i],0)==0:
            count_dict[arr[i]]=1 
        else: 
            count_dict[arr[i]]+=1
    req_occurr=len(arr)//2 
    for value,count in count_dict.items(): 
        if count>req_occurr:
            return value 
    else: 
        return "no majority element "
    
print(find_majority([1,2,3,3,4,3,3]))
```

```python
#let see the optimal appraoch for this

def find_majority(arr):
    count=0 
    candid=None 
    for i in arr: 
        if count==0: 
            candid=i
            count=1
        elif candid==i: 
            count+=1 
        elif candid!=i: 
            count-=1 
    if arr.count(candid)>len(arr)//2: #verification
        return candid 
    else: 
        return 'no majority element'
find_majority([1,2,3,3,3])
```

```python
#maximum subarry sum (kadane algorithm) 
def max_subarray_sum(arr):
    max_sum=float('-inf') 
    sum=0 
    for i in range(len(arr)):
        sum+=arr[i] 
        max_sum=max(sum,max_sum) 
        if sum<0:
            sum=0 
    else: 
        return max_sum 
print(max_subarray_sum([2,4,5,-6,-2,4,0]))
print(max_subarray_sum([-2,-5,-3,-1]))
```

```python
#solving the buy and sell stock problem 

def buy_and_sell_stock(prices):
    min_price=float('inf') 
    max_profit=0 
    #our goal is to maximize the profit. 
    for i in prices: 
        min_price=min(i,min_price) 
        max_profit=max(i-min_price,max_profit) 
        
    return max_profit 

print(buy_and_sell_stock([1,2,3,45]))
print(buy_and_sell_stock([4,1,2,4,9,6,4]))
```

```python
#finding the leaders from the array 
def find_leaders(arr): 
    l=len(arr) 
    leaders=list() 
    max_value=arr[l-1] 
    leaders.append(max_value) 
    for i in range(l-2,-1,-1): 
        
        if arr[i]>max_value: 
            leaders.append(arr[i]) 
            max_value=arr[i] 
    return leaders 



print(find_leaders([1,2,9,4,30]))
```

```python
#longest cosecutive sequence in the array 

def long_consecutive_sequence(arr):
    s=set(arr) 
    max_count=1
    for i in arr: 
        if i-1 not in s: 
            count=1 
            while i+1 in s: 
                count+=1 
                i+=1
                s.remove(i)
            max_count=max(max_count,count)
        else: 
            continue 
    
    return max_count 
print(long_consecutive_sequence([3,5,6,7,7,5,2,1]))
```

```python
#set matrix zeros 

#here if any of the element is zero

#then we have to make its entire row and column zero 


# 1st approach 
# space complexity o(n*m) 
# time complexity o(n*m)
import copy
def set_matrix_zeros(matrix):
    
    #firstly we have to findout the number of rows and columns 
    #here i wll use an different array for marking 
    # so the space complexity is o(n2)
    rows=len(matrix) 
    columns=len(matrix[0])
    ans=copy.deepcopy(matrix) #this is shallow copy so it will not work 
    
   
    for i in range(rows):
        for j in range(columns): 
            if matrix[i][j]==0:
                for l in range(rows):
                    ans[l][j]=0 
                for m in range(columns):
                    ans[i][m]=0 
            
    
    return ans
print(set_matrix_zeros([[1,2,3],[2,0,4],[2,3,45]]))
```

```python
# 2nd approach for this problem
# space complexity o(n+m) 
# time complexity o(n*m)


def set_matrix_zeros(matrix):
    rows=len(matrix) 
    columns=len(matrix[0]) 
    zeros_row=set() 
    zeros_col=set() 
    #now we have to findout all the rows and columns with 0s
    for i in range(rows): 
        for j in range(columns): 
            if matrix[i][j]==0: 
                zeros_row.add(i) 
                zeros_col.add(j) 
    #now we have to fill the zeros in the matrix 
    for i in range(rows): 
        for j in range(columns): 
            if i in zeros_row or j in zeros_col: 
                matrix[i][j]=0 
    
    return matrix 
print(set_matrix_zeros([[1,2,3],[2,0,4],[2,3,45]]))
```

```python
# 2sum problem

def two_sum(arr,s): 
    #here we have to check that the sum is exist in the
    # given array or not by 2 values 
    # eg: 2,3,5,6 s=5 ans = 0,1 
    #we have to return the index of the elements
    dic=dict() 
    for i in range(len(arr)): 
        if s-arr[i] in dic: 
            return [i,dic[s-i]] 
         
        dic[arr[i]]=i 
    else: 
        return "Not Existed" 
    
print(two_sum([1,2,4,9],7))
            
    
```

```python
#rotatet_matrix_by_90_degree clockwise 
def rotate_matrix_90_clockwise(matrix): 
    #first transponse 
    #then reverse the rows 
    
    
    rows=len(matrix ) 
    cols=len(matrix[0]) 
    for i in range(rows): 
        for j in range(i+1,cols):
             
            matrix[i][j],matrix[j][i]=matrix[j][i],matrix[i][j]
    
    #now we have to reverse each row 
    for i in range(rows): 
        
        matrix[i].reverse() 
    return matrix 


    
print(rotate_matrix_90_clockwise([[1,2,3],[4,5,6],[7,8,9]]))
```

```python
#lets find the longest sub array with the given sum 

def longest_subarray_with_given_sum(arr,k): 
    l=len(arr) 
    prefix_sum=0 
    d=dict() 
    max_length=float('-inf') 
    for i in range(l): 
        prefix_sum+=arr[i]  
        if prefix_sum==k: 
            max_length=max(max_length,i+1)
        if prefix_sum-k in d: 
            t=i-d[prefix_sum-k] 
            max_length=max(max_length,t) 
        if prefix_sum not in d: 
            d[prefix_sum]=i
        
    if max_length==float('-inf'): #means subarray doesn't exists 
        return 0
    return max_length 

print(longest_subarray_with_given_sum([1,2,3,4,5],9))
```

```python
def three_sum(arr,target): 
    #sorting+2pointer approach 
    
    arr.sort()
    l=len(arr) 
    left=0 
    ans=list() 
    right=l-1 
    for i in range(l-2): 
        if arr[i]==arr[i+1]:
            continue 
        left=i+1 
        while left<right: 
            if arr[left]+arr[right]+arr[i]==target: 
                ans.append([arr[i],arr[left],arr[right]]) 
                left+=1 
                right-=1 
                while left<right and arr[left]==arr[left-1]: 
                    left+=1
                while left<right and arr[right]==arr[right+1]: 
                    right-=1
            elif arr[left]+arr[right]+arr[i]<target:
                left+=1 
            else: 
                right-=1 
    return ans

print(three_sum([-1,0,1,3,-1,-4],0))
        
```

```python
#multiplication of array except self 

def multiplication_of_array_except_itself(arr): 
    mul=1 
   
    count_0s=0 #for more then 1 0's 
    for i in arr: 
        if i==0: 
            count_0s+=1 
            if count_0s==2:
                break 
            continue 
        mul*=i 
    if count_0s>1:
        return [0]*len(arr) 
    for i in range(len(arr)):
        val=arr[i] 
        if arr[i]==0:
            val=1
        
        arr[i]=mul//val 
    return arr 
print(multiplication_of_array_except_itself([1,2,3,5]))
```

```python
#OPTIMAL APPROACH FOR ABOVE PROBLEM WITHOUT DIVISION OPERATOR

def multiplication_except_self(arr): 
    result=[1]*len(arr) 
    prefix=1 
    suffix=1 
    
    for i in range(len(arr)): 
        result[i]=prefix 
        prefix*=arr[i] 
    suffix=1 
    
    for j in range(len(arr)-1,-1,-1): 
        result[j]*=suffix
        suffix*=arr[j] 
    return result

print(multiplication_except_self([1,2,3,4]))
        
```

```python
#finding the lowerBound in the sorted array 

def find_lower_bound(arr,s ): 
    l=len(arr) 
    left=0 
    min_index=float('inf')
    right = l-1 
    while left<=right: 
        mid=left+(right-left)//2 
        
        if arr[mid]>=s: 
            min_index=min(mid,min_index) 
        
        if s<arr[mid]: 
            right=mid-1 
        else: 
            left=mid+1
    if min_index==float('inf'): 
        return "Not Found"
    return min_index 
print(find_lower_bound([1,3,4,6,7],4))
def find_upper_bound(arr,s ): 
    l=len(arr) 
    left=0 
    right = l-1 
    min_index=float('inf')
    while left<=right: 
        mid=left+(right-left)//2 
        
        if arr[mid]>s: 
            min_index=min(mid,min_index) 
        
        if s<arr[mid]: 
            right=mid-1 
        else: 
            left=mid+1 
    if min_index==float('inf'): 
        return "Not Found"
    return min_index 
    
print(find_upper_bound([1,3,4,6,7],4)) 
```
